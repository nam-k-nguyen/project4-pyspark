```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import size, to_date, year, col, rank, desc
from pyspark.sql.types import *
from pyspark.sql.window import Window

import pandas as pd
import os

# spark = SparkSession.builder.master("local")\
#         .config("spark.port.maxRetries", 100)\
#         .config("spark.executor.instances", "6")\
#         .config("spark.executor.cores", "4")\
#         .config("spark.executor.memory", "8G")\
#         .config("spark.driver.memory", "2G")\
#         .config("spark.dynamicAllocation.enabled", "false")\
#         .config("spark.yarn.queue", "Low")\
#         .config("spark.port.maxRetries", 100)\
#         .appName("New_Data_Reader_V0")\
#         .getOrCreate()

spark = SparkSession.builder.master("local").getOrCreate()
```

```python
cin_station_num = "72429793812"
flo_station_num = "99495199999"
stations = [
    (cin_station_num, "Cincinnati"),
    (flo_station_num, "Florida")
]

years   = list(range(2015, 2025))
base_dir = r"C:\Users\namng\OneDrive\Documents\CloudComputingProjects\project4-pyspark\Data"

print("Idx | Year | Station     | Location    | Count")
idx = 1
for station_code, loc in stations:
    for year in years:
        path = os.path.join(base_dir, str(year), f"{station_code}.csv")
        if os.path.exists(path):
            df = (
                spark.read
                     .option("header",      "true")
                     .option("inferSchema", "true")
                     .option("delimiter",   ",")
                     .csv(path)
            )
            cnt = df.count()
            print(f"{idx:>2d}  | {year} | {station_code} | {loc:<11} | {cnt:5d}")
        else:
            print(f"{idx:>2d}  | {year} | {station_code} | {loc:<11} | MISSING")
        idx += 1
```

```python
# 2) File paths
file_dir    = r"C:\Users\namng\OneDrive\Documents\CloudComputingProjects\project4-pyspark\Data"
cin_file    = os.path.join(file_dir, "72429793812.csv")
flo_file    = os.path.join(file_dir, "99495199999.csv")

# 3) Load the two CSVs
schema_opts = {
    "header":      "true",
    "inferSchema": "true",
    "delimiter":   ","
}
df_cin = spark.read.options(**schema_opts).csv(cin_file)
df_flo = spark.read.options(**schema_opts).csv(flo_file)

# 4) Union into one DataFrame and parse date/year
df = (
    df_cin.union(df_flo)
          .withColumn("DATE", to_date(col("DATE"), "yyyy-MM-dd"))
          .withColumn("year", year(col("DATE")))
)

# 5) Hottest day per year
w_hot = Window.partitionBy("year").orderBy(desc("MAX"))
hottest = (
    df
    .withColumn("rnk", rank().over(w_hot))
    .filter(col("rnk") == 1)
    .select("year", "STATION", "NAME", "DATE", "MAX")
    .orderBy("year")
)

print("=== Hottest day each year ===")
hottest.show(10, truncate=False)

# 6) Coldest March day across all years
coldest_march = (
    df
    .filter(month(col("DATE")) == 3)
    .orderBy(col("MIN").asc())
    .select("STATION", "NAME", "DATE", "MIN")
    .limit(1)
)

print("=== Coldest day in March (2015–2024) ===")
coldest_march.show(truncate=False)
```
